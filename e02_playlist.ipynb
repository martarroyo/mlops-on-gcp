## Parte 5 · La clase `Playlist`

Una playlist tiene un nombre, un tiempo máximo y una lista de episodios que va creciendo. También tiene que saber cuánto dura lo que lleva dentro y si le cabe un episodio más.

> Son unos datos (atributos) junto con las funciones que trabajan sobre ellos (métodos). Es una **clase**.

Las duraciones van a ser objetos `timedelta`, para poder sumarlas y compararlas sin pensar en unidades.

### 🔎 Lee el error

Para saber cuánto dura la playlist hay que sumar duraciones. Lo primero que se te ocurre probablemente sea `sum()`:

````python
duraciones = [timedelta(minutes=44), timedelta(minutes=55)]

sum(duraciones)
````

```
TypeError: unsupported operand type(s) for +: 'int' and 'datetime.timedelta'
```

`sum()` empieza a contar desde el número `0` y le va sumando cada elemento. El primer paso es `0 + timedelta(minutes=44)`, y Python no sabe sumar un número con una duración.

La solución es el patrón *acumular* del módulo 3, empezando desde una duración vacía, `timedelta()`:

````python
total = timedelta()
for duracion in duraciones:
    total += duracion

print(total)
````

### La clase, primera versión

El constructor y el primer método ya están escritos:

````python
class Playlist:
    """Una lista de episodios que cabe en un tiempo máximo."""

    def __init__(self, nombre, minutos_maximos):
        self.nombre = nombre
        self.tiempo_maximo = timedelta(minutes=minutos_maximos)
        self.episodios = []

    def duracion_total(self):
        """Devuelve la duración de todos los episodios como un timedelta."""
        total = timedelta()
        for episodio in self.episodios:
            total += timedelta(minutes=episodio["duracion_min"])
        return total
````

Fíjate en dos cosas del constructor:

- Recibe `minutos_maximos` como número y guarda un `timedelta`. Un atributo no tiene por qué ser igual al dato que recibe el constructor.
- `self.episodios = []` no viene de ningún parámetro. Cada playlist nace con su propia lista vacía.

````python
playlist_prueba = Playlist("Prueba", 60)

print(playlist_prueba.nombre)             # atributo
print(playlist_prueba.tiempo_maximo)      # atributo
print(playlist_prueba.episodios)          # atributo
print(playlist_prueba.duracion_total())   # método
````
